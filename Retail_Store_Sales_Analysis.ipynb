```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. LOAD AND INSPECT DATA

df = pd.read_csv('retail_store_sales.csv')

print(df.head())
print(df.tail())
print("Total rows:", df.shape[0])
print("Total columns:", df.shape[1])
print("Column names:", df.columns)
print(df.dtypes)
df.info()
print(df.describe())
print(df.describe(include='object'))

# 2. MISSING VALUES AND DUPLICATES

missing = df.isnull().sum()
print(missing)
missing_percentage = (df.isnull().sum() / len(df)) * 100
print(missing_percentage)

missing_table = pd.DataFrame({
    'Missing Count': missing,
    'Missing Percentage': round(missing_percentage, 2)
})
print(missing_table)

plt.figure(figsize=(10, 5))
sns.barplot(x=missing.index, y=missing.values)
plt.xticks(rotation=45)
plt.ylabel('Number of missing Values')
plt.xlabel('Missing values by column')
plt.tight_layout()
plt.show()

print("Duplicate records:", df.duplicated().sum())
df = df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)
print(df.head())

# 3. UNIQUE VALUES AND CATEGORICAL DISTRIBUTIONS

for column in df.columns:
    print("\n", column)
    print("Unique values:", df[column].nunique())

print(df['Category'].value_counts())
print(df['Payment Method'].value_counts())
print(df['Location'].value_counts())
print(df['Discount Applied'].value_counts(dropna=False))

for column in df.columns:
    print("Distribution:", column)
    print(df[column].value_counts(dropna=False))

print(df.head())
print(df.dtypes)

# 4. DATE CONVERSION AND DATE RANGE

df['Transaction Date'] = pd.to_datetime(
    df['Transaction Date']
)

print(df.dtypes)
print("Latest transaction:", df['Transaction Date'].max())
print("Earliest transaction:", df['Transaction Date'].min())

# 5. HANDLE MISSING ITEM VALUES

print(df[df['Item'].isnull()])

df['Item'] = df['Item'].fillna('Unknown')

print(df['Item'].head(10))
print("Missing prices:", df['Price Per Unit'].isnull().sum())
print(df['Item'].value_counts())

# 6. HANDLE MISSING PRICE, QUANTITY AND TOTAL SPENT

# Fill missing prices using the median price for each item
item_median_price = (
    df.groupby('Item')['Price Per Unit']
    .transform('median')
)

print(item_median_price)

df['Price Per Unit'] = (
    df['Price Per Unit'].fillna(item_median_price)
)

print(df['Price Per Unit'].head(20))
print(df.isnull().sum())

print(df['Quantity'])

df['Quantity'] = df['Quantity'].fillna(
    df['Quantity'].median()
)

print(df.head(20))

# Calculate missing total spending
df['Total Spent'] = df['Total Spent'].fillna(
    df['Quantity'] * df['Price Per Unit']
)

print(df.isnull().sum())

# 7. HANDLE DISCOUNT APPLIED

print(df['Discount Applied'].value_counts(dropna=False))

df['Discount Applied'] = df['Discount Applied'].fillna(False)

print(df['Discount Applied'].value_counts(dropna=False))
print(df.dtypes)

df['Discount Applied'] = df['Discount Applied'].astype(bool)

print(df.dtypes)

# 8. VALIDATE CLEANED DATA

print("Missing values after cleaning:")
print(df.isnull().sum())

print("Data types after cleaning:")
print(df.dtypes)

print("Duplicate records after cleaning:")
print(df.duplicated().sum())

print(df.head())

df['Calulated_total'] = (
    df['Price Per Unit'] * df['Quantity']
)

print(df.head())

df['Differences'] = (
    df['Total Spent'] - df['Calulated_total']
)

print(df['Differences'])
print("Maximum absolute difference:", df['Differences'].abs().max())

invalid_transactions = df[
    df['Differences'].abs() > 0.01
]

print(invalid_transactions)

# 9. FEATURE ENGINEERING

df['Year'] = df['Transaction Date'].dt.year
df['Month'] = df['Transaction Date'].dt.month

print(df.head())

df['Month_name'] = df['Transaction Date'].dt.month_name()
df['Day'] = df['Transaction Date'].dt.day

print(df.head())

df['Day_Name'] = df['Transaction Date'].dt.day_name()

print(df.head())

df['Revenue_Category'] = pd.cut(
    df['Total Spent'],
    bins=[0, 50, 100, 200, 500],
    labels=['Low', 'Medium', 'High', 'Very High']
)

print(df.head())
print("Dataset shape:", df.shape)
print(df['Category'].value_counts())

# 10. UNIVARIATE EDA: CATEGORY DISTRIBUTION

plt.figure(figsize=(10, 5))
sns.countplot(
    data=df,
    y='Category',
    order=df['Category'].value_counts().index
)
plt.title('Number of transactions by category')
plt.xlabel('Number of Transactions')
plt.tight_layout()

# 11. PAYMENT METHOD DISTRIBUTION

plt.figure(figsize=(10, 5))
sns.countplot(
    data=df,
    y='Payment Method',
    order=df['Payment Method'].value_counts().index
)
plt.title('Number of Transactions by Payment Method')
plt.xlabel('Number of Transactions')
plt.tight_layout()
plt.show()

print(df.head())

# 12. DISTRIBUTION OF QUANTITY

plt.figure(figsize=(7, 5))
sns.histplot(
    data=df,
    x='Quantity',
    bins=10,
    kde=True
)
plt.title('Distribution of Quantity')
plt.tight_layout()
plt.show()

# 13. DISTRIBUTION OF TOTAL SPENT

plt.figure(figsize=(7, 5))
sns.histplot(
    data=df,
    x='Total Spent',
    bins=10,
    kde=True
)
plt.title('Distribution of Total Spent')
plt.tight_layout()
plt.show()

print(df.dtypes)

# 14. BOX PLOT: TOTAL SPENT

plt.figure(figsize=(7, 5))
sns.boxplot(y=df['Total Spent'])
plt.title('Boxplot of Total Spent')
plt.tight_layout()
plt.show()

# Lower whisker = Q1 - 1.5 * IQR
# Upper whisker = Q3 + 1.5 * IQR

# 15. BOX PLOT: QUANTITY

plt.figure(figsize=(7, 5))
sns.boxplot(y=df['Quantity'])
plt.title('Boxplot of Quantity')
plt.tight_layout()
plt.show()

print(df.head())

# 16. BIVARIATE EDA: REVENUE BY CATEGORY

category_sale = (
    df.groupby('Category')['Total Spent']
    .sum()
    .sort_values(ascending=False)
)

print(category_sale)

plt.figure(figsize=(7, 5))
sns.barplot(
    y=category_sale.values,
    x=category_sale.index
)
plt.xticks(rotation=90)
plt.title('Total Revenue by category')
plt.ylabel('Revenue')
plt.tight_layout()
plt.show()

# 17. AVERAGE TRANSACTION VALUE BY CATEGORY

avg_cat_sales = (
    df.groupby('Category')['Total Spent']
    .mean()
    .sort_values(ascending=False)
)

print(avg_cat_sales)

plt.figure(figsize=(7, 5))
sns.barplot(
    y=avg_cat_sales.values,
    x=avg_cat_sales.index
)
plt.xticks(rotation=90)
plt.title('Average transaction value by category')
plt.ylabel('Average total spent')
plt.tight_layout()
plt.show()

# 18. PRODUCT ANALYSIS: TOP 10 ITEMS BY REVENUE

item_sales = (
    df.groupby('Item')['Total Spent']
    .sum()
    .sort_values(ascending=False)
)

print(item_sales.head(10))

plt.figure(figsize=(7, 5))
sns.barplot(
    x=item_sales.head(10).values,
    y=item_sales.head(10).index
)
plt.title('Top 10 Items by Revenue')
plt.xlabel('Total Revenue')
plt.tight_layout()
plt.show()

# 19. QUANTITY ANALYSIS: TOP-SELLING PRODUCTS

item_quantity = (
    df.groupby('Item')['Quantity']
    .sum()
    .sort_values(ascending=False)
)

print(item_quantity.head(5))

# 20. DISCOUNT ANALYSIS

discount_analysis = (
    df.groupby('Discount Applied')['Total Spent']
    .agg(['count', 'sum', 'mean'])
)

print(discount_analysis)

plt.figure(figsize=(7, 5))
sns.barplot(
    data=df,
    x='Discount Applied',
    y='Total Spent'
)
plt.title('Total Spent by Discount Applied')
plt.tight_layout()
plt.show()

# 21. TIME SERIES EDA: DAILY REVENUE

daily_sales = (
    df.groupby('Transaction Date')['Total Spent']
    .sum()
)

print(daily_sales.head(20))

plt.figure(figsize=(7, 5))
plt.plot(
    daily_sales.head(20).index,
    daily_sales.head(20).values
)
plt.title('Daily Revenue')
plt.xlabel('Date')
plt.ylabel('Revenue')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()

print(df.head())

# 22. MONTHLY REVENUE

monthly_sales = (
    df.groupby(
        df['Transaction Date'].dt.to_period('M')
    )['Total Spent']
    .sum()
)

print(monthly_sales.tail())

plt.figure(figsize=(14, 6))
plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker='o'
)
plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 23. YEARLY REVENUE

yearly_sales = (
    df.groupby('Year')['Total Spent']
    .sum()
)

print(yearly_sales)

plt.figure(figsize=(8, 5))
sns.barplot(
    x=yearly_sales.index,
    y=yearly_sales.values
)
plt.title('Yearly Revenue')
plt.xlabel('Year')
plt.ylabel('Revenue')
plt.tight_layout()
plt.show()

print(df.head())

# 24. DAY-OF-WEEK ANALYSIS

day_sales = (
    df.groupby('Day_Name')['Total Spent']
    .mean()
)

day_order = [
    'Monday',
    'Tuesday',
    'Wednesday',
    'Thursday',
    'Friday',
    'Saturday',
    'Sunday'
]

day_sales = day_sales.reindex(day_order)

print(day_sales)

plt.figure(figsize=(10, 5))
sns.barplot(
    x=day_sales.index,
    y=day_sales.values
)
plt.title('Revenue by Day of Week')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 25. CORRELATION ANALYSIS

num_cols = [
    'Price Per Unit',
    'Quantity',
    'Total Spent'
]

correlation = df[num_cols].corr()

print(correlation)

# 26. CORRELATION HEATMAP

plt.figure(figsize=(7, 5))
sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm'
)
plt.title('Correlation Matrix')
plt.tight_layout()
plt.show()

# 27. CATEGORY AND LOCATION ANALYSIS

Category_Location = pd.pivot_table(
    df,
    values='Total Spent',
    index='Category',
    columns='Location',
    aggfunc='sum'
)

print(Category_Location)

plt.figure(figsize=(7, 5))
sns.heatmap(
    Category_Location,
    annot=True,
    cmap='coolwarm',
    fmt='.0f'
)
plt.title('Revenue by Category and Location')
plt.tight_layout()
plt.show()

# 28. CATEGORY AND PAYMENT METHOD ANALYSIS

Category_PM = pd.pivot_table(
    df,
    values='Total Spent',
    index='Category',
    columns='Payment Method',
    aggfunc='sum'
)

print(Category_PM)

plt.figure(figsize=(7, 5))
sns.heatmap(
    Category_PM,
    annot=True,
    cmap='coolwarm',
    fmt='.0f'
)
plt.title('Revenue by Category and Payment Method')
plt.tight_layout()
plt.show()

# 29. CUSTOMER ANALYSIS

print(df.head())
print("Number of unique customers:", df['Customer ID'].nunique())

# 30. CUSTOMER SEGMENTATION

customer_summary = (
    df.groupby('Customer ID').agg(
        Total_Spend=('Total Spent', 'sum'),
        Avg_Spend=('Total Spent', 'mean'),
        Transactions=('Transaction ID', 'count'),
        Total_Quantity=('Quantity', 'sum')
    )
    .sort_values('Total_Spend', ascending=False)
)

print(customer_summary)

customer_summary['Customer_Segment'] = pd.cut(
    customer_summary['Total_Spend'],
    bins=[0, 50000, 65000, 68000, np.inf],
    labels=['Low Value', 'Medium Value', 'High Value', 'VIP']
)

print(customer_summary)

# 31. VISUALIZE CUSTOMER SEGMENTS

plt.figure(figsize=(7, 5))
sns.countplot(
    data=customer_summary,
    x='Customer_Segment'
)
plt.title('Customer Segmentation')
plt.xlabel('Customer Segment')
plt.ylabel('Number of Customers')
plt.tight_layout()
plt.show()

# 32. FINAL CLEANED DATASET

print("Final dataset shape:", df.shape)
print("\nFinal missing-value count:")
print(df.isnull().sum())

print("\nFinal data types:")
print(df.dtypes)

print("\nFinal dataset preview:")
print(df.head())

# Optional: save the cleaned dataset
# df.to_csv('retail_store_sales_cleaned.csv', index=False)
```
